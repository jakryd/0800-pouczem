### Functions

> A function is a block of code which only runs when it is called. You can pass data, known as parameters, into a function. A function can return data as a result.

```python
def func(b):
    print(f'Your {b} function!')
    
func(4)
func(2)
func('a')
```

#### Simple functions

```python
def add(a, b):
    return a + b
```

```python
c = add(4, 3)
```

```python
d = c + add(5, 5)
```

```python
d
```

```python
def add(lst, power):
    s = 0
    for i in range(len(lst)):
        s += lst[i] ** power[i]
    return s
```

```python
add([1, 2, 3, 4], [0,0,0,0])
```

Python functions are defined using the `def` keyword. For example:

```python
def sign(x):
    if x > 0:
        return 'positive'
    elif x < 0:
        return 'negative'
    else:
        return 'zero'

for x in [-1, 0, 1, 10, -10]:
    print(sign(x))
```

#### Optional arguments

We will often define functions to take optional keyword arguments, like this:

```python
def hello(arg, name="Jake", loud=False):
    if loud:
        print('HELLO, {}'.format(name.upper()))
    else:
        print('Hello, {}!'.format(name))

hello(10, name="Mat", loud=True)
```

####  `return` statement

By default (if not return explicitly), functions return `None`.

Note the syntax to define a function:

> the `def` keyword is followed by the function’s name, then the arguments of the function are given between parentheses followed by a colon, the function body and return object for optionally returning values.

`return` can be used to return several things:

```python
def return_values(arg_list=[1, 2, 3, 4]):
    a, b, c, d = [arg ** 2 for arg in arg_list]
    return a, b, c, d

ret = return_values([2, 1, 4, 1])

print(type(ret))
```

```python
# key: name, val: age
d = {"Jake": 4, "Matt": 2, "Bob": 30}

def print_dict(d: dict, it: str = "keys") -> str:
    if not isinstance(d, dict):
        raise AttributeError(f"Pass a dict, got {type(d)}")

    info = ""
    
    if it == "keys":
        for name in d.keys():
            info += f"Person {name} of age {d[name]}\n"
            
    elif it == "items":
        for name, age in d.items():
            info += f"Person {name} of age {age}\n"
            
    else:
        print(f"Got wrong it={it}. Select from 'keys' or 'items'")

    return info
```

```python
info = print_dict(d, it="keys")
print(info)
```

```python
d = {"Jake": 5, "Keks": 20}

def func1(dic: dict, method: str="items") -> str:
    info = ""
    
    if not isinstance(dic, dict):
        print("Variable 'dic' is not a dictionary")
        return info
    
    if method == "items":
        for name, age in dic.items():
            # print(f"Person {name}, age {age}")
            info += f"Person {name}, age {age}\n"
            
    elif method == "keys":
        for name in dic.keys():
            # print(f"Person {name}, age {dic[name]}")
            info += f"Person {name}, age {dic[name]}\n"
            
    else:
        print(f"Unknown method: {method}. Please pass 'keys' or 'items'")
        
    return info

info = func1(dic=d, method="keys")
print(info)
```

#### Variable number of parameters

You can implement a function that takes a variable number of parameters:

```python
def variable_args(name, *args, **kwargs):
    print('name is', name)
    print('args is', args)
    print('kwargs is', kwargs)
```

```python
variable_args(name='Adam', name_2="John")
```

```python
variable_args(1, 2, 3, name="str", name_1="sss")
```

```python
variable_args(1, 2, 4, arg=3, arg2='asas')
```

```python
def func(method="xxx", name="sss", *args, **kwargs):
    print(method)
    print(name)
    print('args is', args)
    print('kwargs is', kwargs)
```

Write a function that displays the n first terms of the Fibonacci sequence, defined by:

$\left\{ \begin{array}{ll} U_{0} = 0 \\ U_{1} = 1 \\ U_{n+2} = U_{n+1} + U_{n} \end{array} \right.$


```python
def fib(n):
    f_new = 1; f_mid = 1; f_old = 1
    print(1)
    print(1)
    for i in range(n):
        f_new = f_old + f_mid
        print(f_new)
        f_old = f_mid
        f_mid = f_new
```
